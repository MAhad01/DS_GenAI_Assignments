```python
# 1  How do you create a simple perceptron for basic binary classification?

import numpy as np


X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])  # AND logic gate

weights = np.random.rand(2)
bias = np.random.rand()

def step_function(z):
    return 1 if z >= 0 else 0

def perceptron(X, weights, bias):
    output = []
    for x in X:
        linear_output = np.dot(x, weights) + bias
        output.append(step_function(linear_output))
    return output

output = perceptron(X, weights, bias)
print("Output:", output)
```

```python
# 2 How can you build a neural network with one hidden layer using Keras?

from keras.models import Sequential
from keras.layers import Dense


model = Sequential()

model.add(Dense(4, input_dim=2, activation='relu'))  # 4 neurons in hidden layer

model.add(Dense(1, activation='sigmoid'))  # Binary classification

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

model.fit(X, y, epochs=100, verbose=1)
```

```python
# 3 How do you initialize weights using the Xavier (Glorot) initialization method in Keras?

from keras.initializers import glorot_uniform

model = Sequential()

model.add(Dense(4, input_dim=2, activation='relu', kernel_initializer=glorot_uniform()))

model.add(Dense(1, activation='sigmoid', kernel_initializer=glorot_uniform()))


model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# 4  How can you apply different activation functions in a neural network in Keras?


model.add(Dense(4, input_dim=2, activation='relu'))

model.add(Dense(4, activation='tanh'))

model.add(Dense(1, activation='sigmoid'))
```

```python
# 5  How do you add dropout to a neural network model to prevent overfitting?

from keras.layers import Dropout

model = Sequential()
model.add(Dense(4, input_dim=2, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(1, activation='sigmoid'))
```

```python
# 6 How do you manually implement forward propagation in a simple neural network?

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def forward_propagation(X, weights1, bias1, weights2, bias2):

    z1 = np.dot(X, weights1) + bias1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, weights2) + bias2
    a2 = sigmoid(z2)
    return a2
```

```python
# 7 How do you add batch normalization to a neural network model in Keras?

from keras.layers import BatchNormalization

model = Sequential()
model.add(Dense(4, input_dim=2, activation='relu'))
model.add(BatchNormalization())
model.add(Dense(1, activation='sigmoid'))
```

```python
# 8 How can you visualize the training process with accuracy and loss curves?

import matplotlib.pyplot as plt

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
history = model.fit(X, y, epochs=10, verbose=1, validation_split=0.2)

# Plot accuracy
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()

# Plot loss
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.show()
```

```python
# 9  How can you use gradient clipping in Keras to control the gradient size and prevent exploding gradients?

from keras.optimizers import Adam

# Compile the model with gradient clipping
optimizer = Adam(clipvalue=0.5)
model.compile(loss='binary_crossentropy', optimizer=optimizer, metrics=['accuracy'])
```

```python
# 10 1 How can you create a custom loss function in Keras?

from keras import backend as K
from keras.losses import Loss

def custom_loss(y_true, y_pred):
    return K.mean(K.square(y_pred - y_true))

model.compile(loss=custom_loss, optimizer='adam', metrics=['accuracy'])
```

```python
# 11 1 How can you visualize the structure of a neural network model in Keras?

from keras.utils import plot_model

plot_model(model, to_file='model.png', show_shapes=True, show_layer_names=True)
```
