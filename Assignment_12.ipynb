```python
# 3. How do you perform inference on an online image using the Faster RCNN model and print the predictions?

import torch
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.transforms import functional as F
from PIL import Image
import requests
# Import BytesIO from the io module
from io import BytesIO

# Load Faster RCNN model
model = fasterrcnn_resnet50_fpn(pretrained=True)
model.eval()

# Load image from URL
url = 'https://ichef.bbci.co.uk/images/ic/480xn/p02ntsq0.jpg.webp'
response = requests.get(url)
image = Image.open(BytesIO(response.content))

# Preprocess the image
image_tensor = F.to_tensor(image).unsqueeze(0)

# Perform inference
with torch.no_grad():
    predictions = model(image_tensor)

# Print predictions
print(predictions)
```

```python
# 5. How do you display bounding boxes for the detected objects in an image using Faster RCNN?

import torch
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.transforms import functional as F
from PIL import Image, ImageDraw

# Load Faster RCNN
model = fasterrcnn_resnet50_fpn(pretrained=True)
model.eval()

# Load and preprocess the image
image = Image.open('/content/sample_data/child.jpeg')
image_tensor = F.to_tensor(image).unsqueeze(0)

# Perform inference
with torch.no_grad():
    predictions = model(image_tensor)

# Display bounding boxes
draw = ImageDraw.Draw(image)
for box in predictions[0]['boxes']:
    draw.rectangle(box.tolist(), outline='red', width=2)

image.show()
```

```python
# 9. How do you perform inference on multiple images from a local folder using Faster RCNN and display the bounding boxes for each?

import os
from PIL import Image, ImageDraw
import torch
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.transforms import functional as F

# Load Faster RCNN
model = fasterrcnn_resnet50_fpn(pretrained=True)
model.eval()

# Folder with images
folder_path = 'images_folder'
output_folder = 'output_folder'
os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(folder_path):
    if filename.endswith('.jpg'):  # Process only JPEG files
        image_path = os.path.join(folder_path, filename)
        image = Image.open(image_path)
        image_tensor = F.to_tensor(image).unsqueeze(0)

        # Inference
        with torch.no_grad():
            predictions = model(image_tensor)

        # Draw bounding boxes
        draw = ImageDraw.Draw(image)
        for box in predictions[0]['boxes']:
            draw.rectangle(box.tolist(), outline='red', width=2)

        # Save output
        image.save(os.path.join(output_folder, filename))
```

```python
# 10. How do you visualize the confidence scores alongside the bounding boxes for detected objects using Faster RCNN?

from PIL import Image, ImageDraw
import torch
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.transforms import functional as F

# Load Faster RCNN
model = fasterrcnn_resnet50_fpn(pretrained=True)
model.eval()

# Load and preprocess image
image = Image.open('image.jpg')
image_tensor = F.to_tensor(image).unsqueeze(0)

# Perform inference
with torch.no_grad():
    predictions = model(image_tensor)

# Draw bounding boxes with confidence scores
draw = ImageDraw.Draw(image)
for box, score in zip(predictions[0]['boxes'], predictions[0]['scores']):
    draw.rectangle(box.tolist(), outline='red', width=2)
    draw.text((box[0], box[1]), f'{score:.2f}', fill='red')

image.show()
```
