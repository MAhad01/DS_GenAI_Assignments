```python
# 2. How do you perform inference with Detectron2 using an online image?

import torch
from detectron2.engine import DefaultPredictor
from detectron2.config import get_cfg
from detectron2 import model_zoo
import cv2
import requests
from PIL import Image
from io import BytesIO
import numpy as np

# Load the Detectron2 model configuration and weights
cfg = get_cfg()
cfg.merge_from_file(model_zoo.get_config_file("COCO-Detection/faster_rcnn_R_50_FPN_3x.yaml"))
cfg.MODEL.WEIGHTS = model_zoo.get_checkpoint_url("COCO-Detection/faster_rcnn_R_50_FPN_3x.yaml")
cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = 0.5
cfg.MODEL.DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
predictor = DefaultPredictor(cfg)

# Load an image from a URL
url = "https://media-cdn.tripadvisor.com/media/photo-s/15/e0/b9/c2/the-view-from-a-little.jpg"
response = requests.get(url)
img = Image.open(BytesIO(response.content)).convert("RGB")
img = np.array(img)

# Perform inference
outputs = predictor(img)

# Display results
v = Visualizer(img[:, :, ::-1], MetadataCatalog.get(cfg.DATASETS.TRAIN[0]), scale=1.2)
out = v.draw_instance_predictions(outputs["instances"].to("cpu"))
cv2.imshow('Detected Objects', out.get_image()[:, :, ::-1])
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# 12. How can you apply a color filter (e.g., red filter) to an image?

import cv2
import numpy as np

# Load an image
image = cv2.imread('path_to_image.jpg')

# Apply a red filter
red_filter = np.zeros_like(image)
red_filter[:, :, 2] = 255  # Set the red channel to maximum

# Blend the original image with the red filter
red_image = cv2.addWeighted(image, 0.5, red_filter, 0.5, 0)

# Display the image
cv2.imshow('Red Filtered Image', red_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
